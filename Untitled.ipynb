```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("SampleSuperstore.csv", encoding='latin1')

df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df.info()
```

```python
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales = monthly_sales.reset_index()

monthly_sales['Order Date'] = monthly_sales['Order Date'].astype(str)

monthly_sales.head()
```

```python
monthly_sales['Month_Number'] = range(1, len(monthly_sales) + 1)

monthly_sales.head()
```

```python
X = monthly_sales[['Month_Number']]

y = monthly_sales['Sales']
```

```python
model = LinearRegression()

model.fit(X, y)

print("Model Trained Successfully")
```

```python
monthly_sales['Predicted_Sales'] = model.predict(X)

monthly_sales.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(monthly_sales['Month_Number'], monthly_sales['Sales'], label='Actual Sales')

plt.plot(monthly_sales['Month_Number'], monthly_sales['Predicted_Sales'], label='Predicted Sales')

plt.xlabel("Month Number")
plt.ylabel("Sales")
plt.title("Sales Forecasting")

plt.legend()

plt.show()
```
